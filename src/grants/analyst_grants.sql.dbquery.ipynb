```sql
-- Group is created in the account console (or via SCIM/Terraform). Example name: quest_analysts
GRANT USE CATALOG ON CATALOG rearc_quest TO `quest_analysts`;
GRANT USE SCHEMA ON SCHEMA rearc_quest.gold TO `quest_analysts`;
GRANT SELECT ON SCHEMA rearc_quest.gold TO `quest_analysts`;
-- Deliberately no grants on raw, bronze, silver, nor on the Volume.
-- Optional: expose a stable, documented interface instead of the tables themselves
CREATE OR REPLACE VIEW rearc_quest.gold.v_series_best_year
COMMENT 'Stable analyst interface for best-year-per-series'
AS SELECT series_id, series_label, sector_name, measure_text, best_year, best_year_value,
is_complete_year
FROM rearc_quest.gold.gold_series_best_year;
GRANT SELECT ON VIEW rearc_quest.gold.v_series_best_year TO `quest_analysts`;
-- Verify from a user in the group:
SHOW GRANTS ON SCHEMA rearc_quest.gold;
```
